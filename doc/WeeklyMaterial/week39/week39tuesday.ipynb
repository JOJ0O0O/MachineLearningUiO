### Step 4: the LLM appendix

The rules are in the [guidelines](https://github.com/EducationalMaterialUiO/MachineLearningUiO/blob/main/LLM_Usage_Declaration_Guidelines.md);
in short:

* LLMs (ChatGPT, Claude, Copilot, Gemini, ...) are **permitted and
  encouraged**. Using one does not reduce your responsibility: every result,
  derivation, code block and argument must be something you understand, can
  defend, and have verified.
* Declare **significant** use — content, logic or structure that appears in
  the submission. Using an LLM as a tutor and then writing in your own words
  is not declared; a literature search whose citations you verified is not
  declared. Concealment, not use, is the integrity violation.
* Add an appendix **"Use of AI/LLM Tools"** with: A.1 the tools (name,
  version or access date, access method); A.2 one contribution level per
  *section* of the report; A.3 one level per *code file*. Level 3 text and
  level 3–4 code get a sentence on the prompt and on what you changed.
* In the code: a `LLM-assisted` tag in the docstring of every function at
  level 2 or above; an inline comment for shorter LLM-generated blocks; a
  markdown note above assisted notebook cells.

| Text level | Meaning |
|---|---|
| 0 None | no LLM assistance |
| 1 Language only | spell-check, grammar, minor phrasing |
| 2 Editorial | LLM rewrote or restructured your sentences; you supplied the ideas and checked the result |
| 3 Generative | LLM drafted substantial text from your prompts; you revised, verified and own it |

| Code level | Meaning |
|---|---|
| 0 None | no LLM assistance |
| 1 Debugging | LLM identified a bug; you implemented the fix |
| 2 Snippet | LLM provided a function, loop or short block; you integrated and tested it |
| 3 Skeleton | LLM generated the structure of a script/class; you filled in the logic |
| 4 Substantial | LLM wrote most of a file; you adapted, tested and commented it |

Fill in the two tables below for **what you have actually done so far** — the
abstract and introduction you just wrote, and the code you have. The helper
prints them as LaTeX and as markdown; paste the LaTeX into your report.

````python
def llm_appendix(tools, text_rows, code_rows, fmt="latex"):
    """Render the 'Use of AI/LLM Tools' appendix as LaTeX (default) or markdown.

    tools:     list of strings, e.g. ["Claude (claude.ai, Sept-Oct 2026)"]
    text_rows: list of (section, level 0-3, note)
    code_rows: list of (file, level 0-4, description)
    """
    if fmt == "latex":
        out = ["\\section*{Appendix A: Use of AI/LLM tools}",
               "\\textbf{A.1 Tools used.} " + ("; ".join(tools) if tools else "None") + ".",
               "", "\\textbf{A.2 Text writing and editing.}",
               "\\begin{tabular}{llp{7cm}}", "Section & Level & Notes\\\\ \\hline"]
        out += [f"{s} & {l} & {n}\\\\" for s, l, n in text_rows]
        out += ["\\end{tabular}", "", "\\textbf{A.3 Code generation and assistance.}",
                "\\begin{tabular}{llp{7cm}}", "File / notebook & Level & Description\\\\ \\hline"]
        out += [f"\\texttt{{{f}}} & {l} & {d}\\\\" for f, l, d in code_rows]
        out += ["\\end{tabular}"]
    else:
        out = ["### Appendix A: Use of AI/LLM tools", "",
               "**A.1 Tools used:** " + ("; ".join(tools) if tools else "None"), "",
               "**A.2 Text**", "", "| Section | Level | Notes |", "|---|---|---|"]
        out += [f"| {s} | {l} | {n} |" for s, l, n in text_rows]
        out += ["", "**A.3 Code**", "", "| File | Level | Description |", "|---|---|---|"]
        out += [f"| `{f}` | {l} | {d} |" for f, l, d in code_rows]
    return "\n".join(out)


# Fill these in for YOUR project -- the entries below are an example only
tools = ["Claude (claude.ai, September 2026)",
         "GitHub Copilot (VS Code, throughout)"]
text_rows = [("Abstract", 1, "grammar check"),
             ("Introduction", 2, "two paragraphs rephrased for clarity; ideas and structure ours"),
             ("Theory / Methods", 0, "--"),
             ("Results", 0, "--"),
             ("Conclusion", 1, "minor language corrections")]
code_rows = [("regression.py", 0, "written independently"),
             ("optimisers.py", 2, "Copilot suggested the Adam bias-correction lines; "
                                 "tested against the closed-form OLS solution"),
             ("resampling.py", 1, "Claude found an off-by-one error in the k-fold split"),
             ("plots.ipynb", 1, "Copilot debugged an axis-indexing bug")]

print(llm_appendix(tools, text_rows, code_rows, fmt="latex"))
print()
print(llm_appendix(tools, text_rows, code_rows, fmt="markdown"))
````
